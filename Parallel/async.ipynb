```python
import asyncio
import time

# ----------------------------
# One bin pipeline (like one field - waiting for I/0)
# ----------------------------
async def process_bin(bin_id, npages=4):
    for page in range(npages):
        await asyncio.sleep(0.5)
    print(f"Bin {bin_id}: done")

# ----------------------------
# Driver (parallel or serial)
# ----------------------------
async def run_bins(nbins=6, parallel=False):
    start = time.perf_counter()

    bins = [process_bin(i) for i in range(nbins)]

    if parallel:
        await asyncio.gather(*bins)
    else:
        for b in bins:
            await b

    elapsed = time.perf_counter() - start
    print(f"\nElapsed: {elapsed:.2f}s (parallel={parallel})\n")


# ----------------------------
# Run both modes
# ----------------------------
print("SERIAL (parallel=False)")
await run_bins(parallel=False)

```

```python
print("PARALLEL (parallel=True)")
await run_bins(parallel=True)
```
